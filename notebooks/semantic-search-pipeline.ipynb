```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
import json

train_path = "/kaggle/input/datasets/pragathigayatri/cuad-dataset/train_separate_questions.json"

with open(train_path, "r") as f:
    train_data = json.load(f)
```

```python
clauses = []

for i in range(5):
    clause = train_data["data"][i]["paragraphs"][0]["context"]
    clauses.append(clause)
```

```python
embeddings = embedding_model.encode(clauses)
```

```python
vector_db = {}

for i in range(len(clauses)):
    vector_db[f"contract_{i}"] = embeddings[i]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
query_clause = "This agreement ends if payment obligations are not met"

print(query_clause)
```

```python
query_embedding = embedding_model.encode([query_clause])

print("Query embedding created")
```

```python
similarities = []

for i in range(len(embeddings)):
    score = cosine_similarity(
        [query_embedding[0]],
        [embeddings[i]]
    )[0][0]
    
    similarities.append(score)

print(similarities)
```

```python
best_match = similarities.index(max(similarities))

print("Most similar contract:", best_match)
print("Similarity score:", max(similarities))
```

```python
import pickle

with open("week3_vector_db.pkl", "wb") as f:
    pickle.dump(vector_db, f)

print("Vector DB saved successfully")
```

```python
from sentence_transformers import SentenceTransformer
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
import json
train_path = "/kaggle/input/datasets/pragathigayatri/cuad-dataset/train_separate_questions.json"

with open(train_path, "r") as f:
    train_data = json.load(f)
```

```python
clauses = []
for i in range(5):
    clause = train_data["data"][i]["paragraphs"][0]["context"]
    clauses.append(clause)
```

```python
embeddings = embedding_model.encode(clauses)
```

```python
vector_db = {}
for i in range(len(clauses)):
    vector_db[f"contract_{i}"] = embeddings[i]
```

```python
query_clause = "This agreement ends if payment obligations are not met"

print(query_clause)
```

```python
query_embedding = embedding_model.encode([query_clause])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarities = []

for i in range(len(embeddings)):
    score = cosine_similarity(
        [query_embedding[0]],
        [embeddings[i]]
    )[0][0]

    similarities.append(score)

print(similarities)
```

```python
top_matches = sorted(
    enumerate(similarities),
    key=lambda x: x[1],
    reverse=True
)[:3]

print(top_matches)
```

```python
for contract_id, score in top_matches:
    print("Contract:", contract_id,
          "| Similarity:", score)
```

```python
import pickle

# Save stored contract vectors
with open("/kaggle/working/week3_vector_db.pkl", "wb") as f:
    pickle.dump(vector_db, f)

# Save embeddings array
with open("/kaggle/working/week3_embeddings.pkl", "wb") as f:
    pickle.dump(embeddings, f)

print("Saved successfully")
```
