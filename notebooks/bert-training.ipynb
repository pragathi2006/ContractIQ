```python
train_data = [
    ("Payment must be completed within 30 days", "Payment"),
    ("Either party may terminate this agreement", "Termination"),
    ("Company information shall remain confidential", "Confidentiality"),
    ("The company accepts full legal liability", "Liability")
]

print(train_data)
```

```python
label_map = {
    "Payment": 0,
    "Termination": 1,
    "Confidentiality": 2,
    "Liability": 3
}

encoded_data = []

for text, label in train_data:
    encoded_data.append((text, label_map[label]))

print(encoded_data)
```

```python
texts = [item[0] for item in encoded_data]

encodings = tokenizer(
    texts,
    padding=True,
    truncation=True,
    return_tensors="pt"
)

print(encodings)
```

```python
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=4
)

print(model)
```

```python
import torch

labels = torch.tensor([item[1] for item in encoded_data])

print(labels)
```

```python
outputs = model(
    input_ids=encodings["input_ids"],
    attention_mask=encodings["attention_mask"],
    labels=labels
)

print(outputs.loss)
```

```python
from torch.optim import AdamW

optimizer = AdamW(model.parameters(), lr=5e-5)

loss = outputs.loss

loss.backward()

optimizer.step()

print("Training step completed")
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=4
)
```

```python
import torch

labels = torch.tensor([0,1,2,3])
```

```python
print(model)
```

```python
from torch.optim import AdamW

optimizer = AdamW(model.parameters(), lr=5e-5)

for epoch in range(3):

    outputs = model(
        input_ids=encodings["input_ids"],
        attention_mask=encodings["attention_mask"],
        labels=labels
    )

    loss = outputs.loss

    loss.backward()

    optimizer.step()

    optimizer.zero_grad()

    print("Epoch:", epoch+1, "Loss:", loss.item())
```

```python
texts = [
    "Payment must be completed within 30 days",
    "Either party may terminate agreement anytime",
    "Company information must remain confidential",
    "The company accepts legal liability"
]
```

```python
encodings = tokenizer(
    texts,
    padding=True,
    truncation=True,
    return_tensors="pt"
)
```

```python
print(encodings)
```

```python
from torch.optim import AdamW

optimizer = AdamW(model.parameters(), lr=5e-5)

for epoch in range(3):

    outputs = model(
        input_ids=encodings["input_ids"],
        attention_mask=encodings["attention_mask"],
        labels=labels
    )

    loss = outputs.loss

    loss.backward()

    optimizer.step()

    optimizer.zero_grad()

    print("Epoch:", epoch+1, "Loss:", loss.item())
```

```python
predictions = outputs.logits.argmax(dim=1)

correct = (predictions == labels).sum().item()

accuracy = correct / len(labels)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import precision_score, recall_score

precision = precision_score(
    labels.numpy(),
    predictions.numpy(),
    average="macro"
)

recall = recall_score(
    labels.numpy(),
    predictions.numpy(),
    average="macro"
)

print("Precision:", precision)
print("Recall:", recall)
```

```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

qa_model = AutoModelForQuestionAnswering.from_pretrained(
    "bert-base-uncased"
)

print("QA Model Loaded")
```

```python
question = data["data"][0]["paragraphs"][0]["qas"][0]["question"]

print(question)
```

```python
context = data["data"][0]["paragraphs"][0]["context"]

question = data["data"][0]["paragraphs"][0]["qas"][0]["question"]

inputs = tokenizer(
    question,
    context,
    return_tensors="pt",
    truncation=True
)

print(inputs.keys())
```

```python
outputs = qa_model(**inputs)

print(outputs)
```

```python
answer_start = outputs.start_logits.argmax()

answer_end = outputs.end_logits.argmax() + 1

answer = tokenizer.convert_tokens_to_string(
    tokenizer.convert_ids_to_tokens(
        inputs["input_ids"][0][answer_start:answer_end]
    )
)

print("Predicted Answer:", answer)
```

```python
sample = data["data"][0]["paragraphs"][0]["qas"][0]

print("Question:", sample["question"])
print("Answer:", sample["answers"][0]["text"])
print("Answer Start:", sample["answers"][0]["answer_start"])
```

```python
sample = data["data"][0]["paragraphs"][0]["qas"][0]

question = sample["question"]

answer = sample["answers"][0]["text"]

answer_start = sample["answers"][0]["answer_start"]

answer_end = answer_start + len(answer)

print(question)
print(answer)
print(answer_start, answer_end)
```

```python
context = data["data"][0]["paragraphs"][0]["context"]

encoding = tokenizer(
    question,
    context,
    return_offsets_mapping=True,
    return_tensors="pt",
    truncation=True
)

print(encoding.keys())
```

```python
offsets = encoding["offset_mapping"][0]

print(offsets[:20])
```

```python
start_char = answer_start
end_char = answer_end

offsets = encoding["offset_mapping"][0]

start_token = None
end_token = None

for idx, (start, end) in enumerate(offsets):

    start = start.item()
    end = end.item()

    if start <= start_char < end:
        start_token = idx

    if start < end_char <= end:
        end_token = idx

print("Start Token:", start_token)
print("End Token:", end_token)
```

```python
import torch

outputs = qa_model(
    input_ids=encoding["input_ids"],
    attention_mask=encoding["attention_mask"],
    start_positions=torch.tensor([start_token]),
    end_positions=torch.tensor([end_token])
)

loss = outputs.loss

print("Loss:", loss.item())
```
